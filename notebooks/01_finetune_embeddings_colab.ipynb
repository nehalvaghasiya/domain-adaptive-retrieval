```python
from pathlib import Path
import torch
from domain_adaptive_retrieval.config.loader import load_config
from domain_adaptive_retrieval.models.device import check_device

DOMAIN = 'finance'  # finance | biomedical | legal | ecommerce | code
CFG = load_config(f'configs/domains/{DOMAIN}.yaml', profile='configs/experiments/t4.yaml')
assert CFG.device == 'cuda:0'  # Important: never use the unindexed 'cuda' string here.
print(check_device(CFG.device, require_gpu=True))
# To do a quick subset run first, use:
# CFG.max_train_pairs = 1000
# CFG.max_val_queries = 50
# CFG.max_test_queries = 50
OUTPUT = Path('outputs') / DOMAIN
```

## 3. Fine-tune and reload on CUDA

```python
from domain_adaptive_retrieval.workflows.train_embedder import train_embedding
from domain_adaptive_retrieval.models.embedder import EmbeddingModel

EMBEDDER_PATH = train_embedding(CFG, OUTPUT / 'embedding')
print('Fine-tuned model:', EMBEDDER_PATH)
loaded = EmbeddingModel(EMBEDDER_PATH, device='cuda:0', max_length=CFG.max_sequence_length)
encoded = loaded.encode(['domain-specific retrieval'], convert_to_tensor=True)
print('Reloaded model inference tensor device:', encoded.device)
assert encoded.device.type == 'cuda'
```

## 4. Evaluate against the frozen base model

```python
# Evaluate the baseline against the trained model using the same held-out labels.
import subprocess, sys
subprocess.run([
    sys.executable, '-m', 'domain_adaptive_retrieval.cli', 'evaluate', 'embedder',
    '--checkpoint', str(EMBEDDER_PATH), '--domain', DOMAIN, '--profile', 't4',
    '--output', str(OUTPUT / 'embedding' / 'heldout_evaluation'),
], check=True)
```
